```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('labno2.jpg', 0)

# Check if image loaded
if img is None:
    print("Error: Image not found!")
else:
    # Negative transformation
    negative = 255 - img

    # Bit-plane slicing
    bit_planes = []
    for i in range(8):
        bit_plane = (img >> i) & 1
        bit_planes.append(bit_plane * 255)

    # Show original and negative
    plt.figure(figsize=(10,5))

    plt.subplot(2,5,1)
    plt.imshow(img, cmap='gray')
    plt.title("Original")

    plt.subplot(2,5,2)
    plt.imshow(negative, cmap='gray')
    plt.title("Negative")

    # Show bit planes
    for i in range(8):
        plt.subplot(2,5,i+3)
        plt.imshow(bit_planes[i], cmap='gray')
        plt.title(f'Bit {i}')

    plt.tight_layout()
    plt.show()
```
